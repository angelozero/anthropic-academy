```python
# Criando o client
from anthropic import Anthropic

client = Anthropic()
```

```python
# Funcões utilitarias
def add_user_message(messages, text):
    user_message = {
        "role": "user",
        "content": text,
    }
    messages.append(user_message)


def add_assistant_message(messages, text):
    assistant_message = {
        "role": "assistant",
        "content": text,
    }
    messages.append(assistant_message)
```

```python
messages = []
system_prompt = """
                    Você é um tutor de matemática paciente.
                    Não responda diretamente às perguntas de um aluno.
                    Guie-os até uma solução passo a passo.
                """

add_user_message(messages=messages, text="Qual o valor de x na equação 5x=3+2")

stream = client.messages.create(
    # Info claude-sonnet-4-0' is deprecated and will reach end-of-life on June 15th, 2026.
    model="claude-opus-4-6",
    max_tokens=1024,
    messages=messages,
    stream=True
)

for event in stream:
    print(event)
```

```python
messages = []
system_prompt = """
                    Você é um tutor de matemática paciente.
                    Não responda diretamente às perguntas de um aluno.
                    Guie-os até uma solução passo a passo.
                """

add_user_message(messages=messages, text="Qual o valor de x na equação 5x=3+2")

with client.messages.create(
    # Info claude-sonnet-4-0' is deprecated and will reach end-of-life on June 15th, 2026.
    model="claude-opus-4-6",
    max_tokens=1024,
    messages=messages,
    stream=True
) as stream:
    for text in stream.text_stream:
        # Print de cada 'chunck' da resposta final ao invés do valor inteiro de uma vez só
        print(text, end="")
        
# Para recuperar a resposta final completa        
stream.get_final_message()
```
