```python
# Carregando variaveis .env
from dotenv import load_dotenv

load_dotenv()
```

```python
# Criando o client
from anthropic import Anthropic

client = Anthropic()
```

```python
# Funcões utilitarias
def add_user_message(messages, text):
    user_message = {
        "role": "user",
        "content": text,
    }
    messages.append(user_message)


def add_assistant_message(messages, text):
    assistant_message = {
        "role": "assistant",
        "content": text,
    }
    messages.append(assistant_message)
```

```python
# Fazendo a requisição


def chat(messages, system_prompt=None):

    params = {
        # Info claude-sonnet-4-0' is deprecated and will reach end-of-life on June 15th, 2026.
        "model": "claude-opus-4-6",
        "max_tokens": 1024,
        "messages": messages,
        "system": system_prompt,
    }

    if system_prompt:
        params["system"] = system_prompt

    # Doc ref: https://platform.claude.com/docs/en/api/python/messages/create
    message = client.messages.create(**params)

    return message.content[0].text
```

```python
messages = []
system_prompt = """
                    Você é um tutor de matemática paciente.
                    Não responda diretamente às perguntas de um aluno.
                    Guie-os até uma solução passo a passo.
                """

add_user_message(messages=messages, text="Qual o valor de x na equação 5x=3+2")

answer = chat(messages=messages, system_prompt=system_prompt)

print(answer)
```
