```python
# Carregando variaveis .env
from dotenv import load_dotenv

load_dotenv()
```

```python
# Criando o client
from anthropic import Anthropic

client = Anthropic()
```

```python
# Funcões utilitarias
def add_user_message(messages, text):
    user_message = {
        "role": "user",
        "content": text,
    }
    messages.append(user_message)
    
def add_assistant_message(messages, text):
    assistant_message = {
        "role": "assistant",
        "content": text,
    }
    messages.append(assistant_message)
```

```python
# Fazendo a requisição

def chat(messages):
    # Doc ref: https://platform.claude.com/docs/en/api/python/messages/create
    message = client.messages.create(
        # Info claude-sonnet-4-0' is deprecated and will reach end-of-life on June 15th, 2026.
        model="claude-opus-4-6",
        max_tokens=1024,
        messages=messages,
    )
    
    return message.content[0].text
```

```python
# Criar uma lista inicial de mensagens
messages = []

# Adicionar a pergunta inicial do usuário: "Defina computação quântica em uma frase"
add_user_message(messages, text="Diga ola")

# Passar a lista de mensagens para o 'chat' para obter uma resposta
answer = chat(messages=messages)

# Pegar a resposta e adicioná-la como uma mensagem do assistente em nossa lista
add_assistant_message(messages=messages, text=answer)

# Adicionar a pergunta de acompanhamento do usuário
add_user_message(messages=messages, text="Diga com meu nome Angelo")

# Chamar o chat novamente com a lista de mensagens para obter uma resposta final
final_answer = chat(messages=messages)

print(final_answer)

```

```python
# Codigo de teste

messages = []

while True:
    user_input = input("> ")
    print(">", user_input)
    
    add_user_message(messages=messages, text=user_input)
    
    answer = chat(messages=messages)
    print(answer)
    
    add_assistant_message(messages=messages, text=answer)
    
    print("---")
    print(answer)
    print("---")
```
